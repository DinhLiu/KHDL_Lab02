## Write Labels to pred.json Files

Creates `pred.json` for each publication containing:
- `partition`: "train", "valid", or "test"
- `label_type`: "manual" or "auto"
- `bib_keys`: list of all BibTeX citation keys (m entries)
- `arxiv_ids`: list of all reference arXiv IDs (n entries)  
- `positive_pairs`: dict mapping bib_key -> arxiv_id for matches (label=1)
- `prediction`: placeholder for model predictions

**Data points per publication**: m × n pairs with binary labels

```python
import re

def load_metadata(pub_path: Path) -> dict:
    """Load metadata.json containing paper info (title, authors, submission_date, revised_dates, venue)."""
    meta_file = pub_path / "metadata.json"
    if meta_file.exists():
        with open(meta_file, 'r', encoding='utf-8') as f:
            return json.load(f)
    return {}


def load_references(pub_path: Path) -> dict:
    """Load references.json - dict with arXiv IDs as keys.
    
    Each value contains: title, authors, submission_date, semantic_scholar_id
    """
    ref_file = pub_path / "references.json"
    if ref_file.exists():
        with open(ref_file, 'r', encoding='utf-8') as f:
            return json.load(f)
    return {}


def load_hierarchy(pub_path: Path) -> dict:
    """Load hierarchy.json containing parsed content and hierarchical structure."""
    hierarchy_file = pub_path / "hierarchy.json"
    if hierarchy_file.exists():
        with open(hierarchy_file, 'r', encoding='utf-8') as f:
            return json.load(f)
    return {}


def get_all_bib_keys(pub_path: Path) -> list:
    """Extract all citation keys from refs.bib (unified BibTeX file)."""
    keys = []
    bib_file = pub_path / "refs.bib"
    
    if not bib_file.exists():
        return keys
    
    try:
        with open(bib_file, 'r', encoding='utf-8', errors='ignore') as f:
            content = f.read()
            # Match BibTeX entry keys: @type{key, or @type{key\n
            bib_matches = re.findall(r'@\w+\s*\{\s*([^,\s]+)\s*,', content)
            keys.extend(bib_matches)
    except Exception:
        pass
    
    return list(set(keys))


def write_pred_json(pub_path: Path, groundtruth: dict, partition: str = "train"):
    """Write pred.json file for reference matching task.
    
    Format (as specified):
    {
        "partition": "test",
        "groundtruth": {
            "bibtex_entry_name_1": "arxiv_id_1",
            "bibtex_entry_name_2": "arxiv_id_2",
            ...
        },
        "prediction": {
            "bibtex_entry_name_1": ["cand_1", "cand_2", "cand_3", "cand_4", "cand_5"]
        }
    }
    
    Args:
        pub_path: Path to publication folder
        groundtruth: Dict of {bibtex_entry_name: arxiv_id} - correct mappings
        partition: "train", "valid", or "test"
    
    Returns:
        Tuple of (pred_file_path, num_bib_entries, num_references, num_pairs)
    """
    # Get statistics for reporting
    bib_keys = get_all_bib_keys(pub_path)
    references = load_references(pub_path)
    m = len(bib_keys)
    n = len(references)
    
    pred_data = {
        "partition": partition,
        "groundtruth": groundtruth,
        "prediction": {key: [] for key in groundtruth.keys()}  # Empty predictions placeholder
    }
    
    pred_file = pub_path / "pred.json"
    with open(pred_file, 'w', encoding='utf-8') as f:
        json.dump(pred_data, f, indent=2, ensure_ascii=False)
    
    return pred_file, m, n, m * n
```

```python
def process_all_labels():
    """Process all manual labels and write to pred.json files.
    
    For each publication with m BibTeX entries and n references:
    - Generates m×n data points (pairs)
    - Each pair has binary label: 1 if in groundtruth, 0 otherwise
    
    Output format:
    {
        "partition": "train/valid/test",
        "groundtruth": {bibtex_entry_name: arxiv_id, ...},
        "prediction": {bibtex_entry_name: [ranked candidate list], ...}
    }
    """
    data_path = BASE_PATH / DATA_FOLDER
    
    if not data_path.exists():
        print(f"Error: Data folder not found: {data_path}")
        return
    
    results = {
        "success": [],
        "failed": [],
        "skipped": []
    }
    
    total_data_points = 0
    total_positive = 0
    
    for pub_id, labels in MANUAL_LABELS.items():
        pub_path = data_path / pub_id
        
        if not pub_path.exists():
            results["skipped"].append((pub_id, "Publication folder not found"))
            continue
        
        if not labels:
            results["skipped"].append((pub_id, "No labels defined"))
            continue
        
        try:
            # Load references to validate arXiv IDs
            references = load_references(pub_path)
            
            # Get all bib keys from refs.bib
            bib_keys = set(get_all_bib_keys(pub_path))
            
            # Validate labels against references and refs.bib
            valid_groundtruth = {}
            invalid_arxiv = []
            invalid_bibkeys = []
            
            for bib_key, arxiv_id in labels.items():
                arxiv_valid = arxiv_id in references
                bibkey_valid = bib_key in bib_keys if bib_keys else True
                
                if arxiv_valid and bibkey_valid:
                    valid_groundtruth[bib_key] = arxiv_id
                elif not arxiv_valid:
                    invalid_arxiv.append((bib_key, arxiv_id))
                elif not bibkey_valid:
                    invalid_bibkeys.append((bib_key, arxiv_id))
            
            if invalid_arxiv:
                print(f"Warning [{pub_id}]: arXiv IDs not in references.json:")
                for bib_key, arxiv_id in invalid_arxiv:
                    print(f"  - {bib_key} -> {arxiv_id}")
            
            if invalid_bibkeys:
                print(f"Warning [{pub_id}]: Citation keys not in refs.bib:")
                for bib_key, arxiv_id in invalid_bibkeys:
                    print(f"  - {bib_key} -> {arxiv_id}")
            
            # Get partition from MANUAL_PARTITIONS (default to train)
            partition = MANUAL_PARTITIONS.get(pub_id, "train")
            
            # Write pred.json
            pred_file, m, n, num_pairs = write_pred_json(
                pub_path, valid_groundtruth, partition=partition
            )
            
            total_data_points += num_pairs
            total_positive += len(valid_groundtruth)
            
            results["success"].append({
                "pub_id": pub_id,
                "partition": partition,
                "m": m,
                "n": n,
                "pairs": num_pairs,
                "groundtruth_count": len(valid_groundtruth),
                "path": str(pred_file)
            })
                
        except Exception as e:
            results["failed"].append((pub_id, str(e)))
    
    return results, total_data_points, total_positive


# Run the labeling process
print("Processing manual labels...")
print("=" * 70)

results, total_pairs, total_positive = process_all_labels()

print("\nSuccessfully written:")
print(f"{'Pub ID':<15} {'Partition':<8} {'m':<5} {'n':<5} {'m×n':<8} {'GT':<6}")
print("-" * 70)
for r in results["success"]:
    print(f"{r['pub_id']:<15} {r['partition']:<8} {r['m']:<5} {r['n']:<5} {r['pairs']:<8} {r['groundtruth_count']:<6}")

if results["failed"]:
    print("\nFailed:")
    for pub_id, error in results["failed"]:
        print(f"  {pub_id}: {error}")

if results["skipped"]:
    print("\nSkipped:")
    for pub_id, reason in results["skipped"]:
        print(f"  {pub_id}: {reason}")

print("\n" + "=" * 70)
print(f"Total: {len(results['success'])} publications processed")
print(f"Total data points (m×n pairs): {total_pairs:,}")
print(f"Total groundtruth mappings: {total_positive:,}")
```

## Verify Written Labels

Check what was written to verify correctness.

```python
def verify_labels():
    """Read back and display written pred.json files."""
    data_path = BASE_PATH / DATA_FOLDER
    
    print("Verifying written pred.json files:")
    print("=" * 70)
    
    partition_stats = {"train": [], "valid": [], "test": []}
    
    for pub_id in MANUAL_LABELS.keys():
        pub_path = data_path / pub_id
        pred_file = pub_path / "pred.json"
        
        if pred_file.exists():
            with open(pred_file, 'r', encoding='utf-8') as f:
                data = json.load(f)
            
            partition = data.get('partition', 'train')
            groundtruth = data.get('groundtruth', {})
            prediction = data.get('prediction', {})
            
            # Get actual counts from files
            bib_keys = get_all_bib_keys(pub_path)
            references = load_references(pub_path)
            m, n = len(bib_keys), len(references)
            
            partition_stats[partition].append({
                "pub_id": pub_id,
                "m": m,
                "n": n,
                "groundtruth_count": len(groundtruth)
            })
            
            print(f"\n{pub_id}/pred.json:")
            print(f"  Partition: {partition}")
            print(f"  BibTeX entries (m): {m}")
            print(f"  References (n): {n}")
            print(f"  Groundtruth mappings: {len(groundtruth)}")
            print(f"  Prediction entries: {len(prediction)}")
            
            # Show sample groundtruth mappings
            if groundtruth:
                print(f"  Sample groundtruth:")
                for bib_key, arxiv_id in list(groundtruth.items())[:3]:
                    print(f"    {bib_key} -> {arxiv_id}")
                if len(groundtruth) > 3:
                    print(f"    ... and {len(groundtruth) - 3} more")
        else:
            print(f"\n[NOT FOUND] {pub_id}/pred.json")
    
    # Summary by partition
    print("\n" + "=" * 70)
    print("PARTITION SUMMARY:")
    for partition, pubs in partition_stats.items():
        if pubs:
            total_gt = sum(p['groundtruth_count'] for p in pubs)
            print(f"\n  {partition.upper()}:")
            print(f"    Publications: {len(pubs)}")
            print(f"    Total groundtruth mappings: {total_gt}")

verify_labels()
```

## Helper: Explore Publication for Labeling

Use this cell to explore a specific publication and identify matches.

```python
def explore_publication(pub_id: str):
    """Explore a publication to help with manual labeling.
    
    Uses the new data structure:
    - metadata.json: paper title, authors, submission_date, revised_dates, venue
    - references.json: arXiv ID -> {title, authors, submission_date, semantic_scholar_id}
    - refs.bib: unified BibTeX entries
    - hierarchy.json: parsed content and hierarchical structure
      - Element IDs format: {pub_id}_{type}_{counter} (e.g., 2411-00230_section_0001)
      - Elements are deduplicated across versions (full-text match)
    """
    data_path = BASE_PATH / DATA_FOLDER / pub_id
    
    if not data_path.exists():
        print(f"Publication not found: {pub_id}")
        return
    
    print(f"Exploring: {pub_id}")
    print("=" * 60)
    
    # Load and display metadata
    metadata = load_metadata(data_path)
    if metadata:
        print(f"\nPaper Metadata:")
        print(f"  Title: {metadata.get('title', 'N/A')}")
        print(f"  Authors: {', '.join(metadata.get('authors', [])[:5])}")
        if len(metadata.get('authors', [])) > 5:
            print(f"           ... and {len(metadata['authors']) - 5} more")
        print(f"  Submission Date: {metadata.get('submission_date', 'N/A')}")
        print(f"  Revised Dates: {metadata.get('revised_dates', [])}")
        print(f"  Venue: {metadata.get('venue', 'N/A')}")
    
    # Load references (arXiv papers)
    references = load_references(data_path)
    print(f"\nCandidate arXiv Papers ({len(references)}):")
    for arxiv_id, info in list(references.items())[:20]:  # Show first 20
        title = info.get('title', info.get('paper_title', 'N/A'))[:60]
        authors = ', '.join(info.get('authors', [])[:3])
        ss_id = info.get('semantic_scholar_id', 'N/A')
        print(f"  {arxiv_id}: {title}...")
        print(f"           by {authors}")
        print(f"           SS ID: {ss_id}")
    
    if len(references) > 20:
        print(f"  ... and {len(references) - 20} more")
    
    # Get citation keys from refs.bib
    bib_keys = get_all_bib_keys(data_path)
    print(f"\nBibTeX Citation Keys from refs.bib ({len(bib_keys)}):")
    for key in sorted(bib_keys)[:30]:  # Show first 30
        print(f"  - {key}")
    
    if len(bib_keys) > 30:
        print(f"  ... and {len(bib_keys) - 30} more")
    
    # Load hierarchy info
    hierarchy = load_hierarchy(data_path)
    if hierarchy:
        elements = hierarchy.get('elements', {})
        hier_versions = hierarchy.get('hierarchy', {})
        element_versions = hierarchy.get('element_versions', {})
        
        print(f"\nHierarchy Structure:")
        print(f"  Total deduplicated elements: {len(elements)}")
        print(f"  Hierarchy versions: {list(hier_versions.keys())}")
        
        # Analyze element ID format
        if elements:
            sample_id = list(elements.keys())[0]
            print(f"  Element ID format: {{pub_id}}_{{type}}_{{counter}}")
            print(f"  Example ID: {sample_id}")
        
        # Count elements by type
        type_counts = {}
        for elem_id in elements.keys():
            # Extract type from ID: {pub_id}_{type}_{counter}
            parts = elem_id.split('_')
            if len(parts) >= 3:
                elem_type = parts[-2]  # Type is second to last part
                type_counts[elem_type] = type_counts.get(elem_type, 0) + 1
        
        if type_counts:
            print(f"\n  Elements by type:")
            for elem_type, count in sorted(type_counts.items(), key=lambda x: -x[1]):
                print(f"    {elem_type}: {count}")
        
        # Show sample elements with their versions
        print(f"\n  Sample elements (with version info):")
        for elem_id, content in list(elements.items())[:5]:
            content_preview = str(content)[:60].replace('\n', ' ')
            versions = element_versions.get(elem_id, ['unknown'])
            print(f"    {elem_id}")
            print(f"      Versions: {versions}")
            print(f"      Content: {content_preview}...")


# Example: explore a publication
# explore_publication("2411-00230")
```

```python
def compute_statistics():
    """Compute statistics for reference matching task."""
    data_path = BASE_PATH / DATA_FOLDER
    
    stats = {
        "manual": {"test": [], "valid": [], "train": []},
        "auto": {"test": [], "valid": [], "train": []}
    }
    
    # Process manual labels
    for pub_id, labels in MANUAL_LABELS.items():
        pub_path = data_path / pub_id
        if pub_path.exists():
            bib_keys = get_all_bib_keys(pub_path)
            references = load_references(pub_path)
            m, n = len(bib_keys), len(references)
            partition = MANUAL_PARTITIONS.get(pub_id, "train")
            
            stats["manual"][partition].append({
                "pub_id": pub_id,
                "m": m,
                "n": n,
                "pairs": m * n,
                "groundtruth": len(labels)
            })
    
    # Display statistics
    print("=" * 70)
    print("REFERENCE MATCHING STATISTICS")
    print("=" * 70)
    
    grand_total_pairs = 0
    grand_total_gt = 0
    
    for label_type in ["manual", "auto"]:
        type_pairs = 0
        type_gt = 0
        
        pubs_exist = any(stats[label_type][p] for p in stats[label_type])
        if not pubs_exist:
            continue
            
        print(f"\n{label_type.upper()} LABELS:")
        print("-" * 50)
        
        for partition in ["train", "valid", "test"]:
            pubs = stats[label_type][partition]
            if pubs:
                pairs = sum(p["pairs"] for p in pubs)
                gt = sum(p["groundtruth"] for p in pubs)
                type_pairs += pairs
                type_gt += gt
                
                print(f"  {partition.upper():8} | {len(pubs)} pubs | {pairs:>8,} pairs | {gt:>5} groundtruth")
        
        if type_pairs > 0:
            total_pubs = sum(len(stats[label_type][p]) for p in stats[label_type])
            print(f"  {'SUBTOTAL':8} | {total_pubs} pubs | {type_pairs:>8,} pairs | {type_gt:>5} groundtruth")
        
        grand_total_pairs += type_pairs
        grand_total_gt += type_gt
    
    if grand_total_pairs > 0:
        print("\n" + "=" * 70)
        print("GRAND TOTAL:")
        print(f"  Total publications: {sum(len(stats[t][p]) for t in stats for p in stats[t])}")
        print(f"  Total data points (m×n pairs): {grand_total_pairs:,}")
        print(f"  Total groundtruth mappings: {grand_total_gt:,}")
        print(f"  Match rate: {100*grand_total_gt/grand_total_pairs:.3f}%")
    
    # Per publication breakdown
    print("\n" + "=" * 70)
    print("PER PUBLICATION BREAKDOWN:")
    print(f"{'Pub ID':<15} {'Type':<8} {'Part':<6} {'m (bib)':<8} {'n (ref)':<8} {'m×n':<10} {'GT':<6}")
    print("-" * 70)
    
    for label_type in ["manual", "auto"]:
        for partition in ["train", "valid", "test"]:
            for p in stats[label_type][partition]:
                print(f"{p['pub_id']:<15} {label_type:<8} {partition:<6} {p['m']:<8} {p['n']:<8} {p['pairs']:<10,} {p['groundtruth']:<6}")

compute_statistics()
```
